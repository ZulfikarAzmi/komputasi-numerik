# Penyelesaian Sistem Persamaan Linear Tiga Variabel (SPLTV)

Program ini digunakan untuk mencari nilai x, y, dan z dari tiga persamaan linear menggunakan Python.


```python

import numpy as np
```

Mengimpor NumPy untuk membantu perhitungan matriks.

```python

# Contoh persamaan:
# x + y + z = 6
# 2x - y + z = 3
# x + 2y - z = 2

A = np.array([
    [1, 1, 1],
    [2, -1, 1],
    [1, 2, -1]
])

B = np.array([6, 3, 2])
```

Matriks A menyimpan koefisien variabel, sedangkan B menyimpan hasil setiap persamaan.

```python

print("Penyelesaian SPLTV")
print("Persamaan 1: x + y + z = 6")
print("Persamaan 2: 2x - y + z = 3")
print("Persamaan 3: x + 2y - z = 2")
```

Menampilkan tiga persamaan yang akan diselesaikan.

```python

if np.linalg.matrix_rank(A) == 3:
    hasil = np.linalg.solve(A, B)
    print("Sistem memiliki solusi tunggal.")
else:
    hasil = None
    print("Sistem tidak memiliki solusi tunggal.")
```

Memeriksa peringkat matriks dan menghitung solusi jika sistem memiliki solusi tunggal.

```python

if hasil is not None:
    print("Nilai x =", round(hasil[0], 2))
    print("Nilai y =", round(hasil[1], 2))
    print("Nilai z =", round(hasil[2], 2))
```

Menampilkan nilai akhir variabel \(x\), \(y\), dan \(z\).

```python

if hasil is not None:
    print("Pemeriksaan hasil:")
    print("Persamaan 1 =", round(sum(hasil), 2))
    print("Persamaan 2 =", round(2*hasil[0] - hasil[1] + hasil[2], 2))
    print("Persamaan 3 =", round(hasil[0] + 2*hasil[1] - hasil[2], 2))
```
