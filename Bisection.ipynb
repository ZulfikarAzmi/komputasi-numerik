# Penyelesaian Persamaan Nonlinear Metode Bisection

Program ini digunakan untuk mencari akar persamaan nonlinear menggunakan Metode Bisection atau metode bagi dua.


```python
import math
```

Mengimpor library math untuk melakukan operasi matematika jika diperlukan.

```python
# Contoh persamaan: f(x) = x^3 - x - 2

def f(x):
    return x**3 - x - 2

# Interval awal
a = 1
b = 2

# Toleransi error
toleransi = 0.0001

# Maksimum iterasi
maks_iterasi = 100
```

Menentukan fungsi nonlinear, batas interval awal \(a\) dan \(b\), toleransi error, serta jumlah iterasi maksimum.

```python

print("Persamaan: f(x) = x^3 - x - 2")
print("Interval awal:", a, "sampai", b)
print("f(a) =", f(a))
print("f(b) =", f(b))

if f(a) * f(b) < 0:
    print("Interval memenuhi syarat metode Bisection.")
else:
    print("Interval tidak memenuhi syarat metode Bisection.")
```

Metode Bisection membutuhkan nilai \(f(a)\) dan \(f(b)\) yang berbeda tanda agar terdapat akar pada interval tersebut.

```python

if f(a) * f(b) < 0:
    for i in range(1, maks_iterasi + 1):
        c = (a + b) / 2

        print(f"Iterasi {i}: a={a:.6f}, b={b:.6f}, "
              f"c={c:.6f}, f(c)={f(c):.6f}")

        if abs(f(c)) < toleransi or (b - a) / 2 < toleransi:
            break

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    akar = c
else:
    akar = None
```

Menghitung titik tengah \(c=(a+b)/2\).



Mengevaluasi nilai \(f(c)\).



Mempersempit interval berdasarkan tanda fungsi.



Menghentikan iterasi jika toleransi terpenuhi atau iterasi maksimum tercapai.

```python

if akar is not None:
    print("\nHasil Penyelesaian:")
    print(f"Akar persamaan = {akar:.6f}")
    print(f"Nilai f(akar) = {f(akar):.6f}")
else:
    print("Akar tidak dapat dihitung karena interval awal tidak memenuhi syarat.")
```

Menampilkan perkiraan akar persamaan dan nilai fungsi pada akar tersebut.
